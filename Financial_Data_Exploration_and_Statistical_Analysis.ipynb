```python
aapl_close = aapl["Close"]["AAPL"]
amzn_close = amzn["Close"]["AMZN"]
nflx_close = nflx["Close"]["NFLX"]

Dstats = pd.DataFrame({
    "Company" : ["AAPL", "AMZN", "NFLX"],
    "Mean": [
        aapl_close.mean(),
        amzn_close.mean(),
        nflx_close.mean()
    ],
    "Median": [
        aapl_close.median(),
        amzn_close.median(),
        nflx_close.median()
    ],
    "Minimum": [
        aapl_close.min(),
        amzn_close.min(),
        nflx_close.min()
    ],
    "Maximum": [
        aapl_close.max(),
        amzn_close.max(),
        nflx_close.max()
    ],
    "Standard Deviation": [
        aapl_close.std(),
        amzn_close.std(),
        nflx_close.std()
    ]
})

Dstats
```

```python
aapl_close = aapl["Close"]["AAPL"]
amzn_close = amzn["Close"]["AMZN"]
nflx_close = nflx["Close"]["NFLX"]

aapl_return = (aapl_close - aapl_close.shift(1)) / aapl_close.shift(1)
amzn_return = (amzn_close - amzn_close.shift(1)) / amzn_close.shift(1)
nflx_return = (nflx_close - nflx_close.shift(1)) / nflx_close.shift(1)


print("AAPL Daily Returns:")
print(aapl_return)

print("\nAMZN Daily Returns:")
print(amzn_return)

print("\nNFLX Daily Returns:")
print(nflx_return)
```

```python
plt.figure(figsize=(12, 6))

plt.plot(aapl_return.index, aapl_return, label="AAPL")
plt.plot(amzn_return.index, amzn_return, label="AMZN")
plt.plot(nflx_return.index, nflx_return, label="NFLX")

plt.xlabel("Date")
plt.ylabel("Daily Return")
plt.title("Daily Stock Returns")
plt.legend()
plt.xticks(rotation=45)

plt.show()
```

1. What was the average daily return?

Average daily return: approximately 0

2. What was the highest daily gain?

Highest daily gain: approximately 0.1

3. What was the largest daily loss?

Largest daily loss: approximately -0.09

```python
volatility = pd.DataFrame({
    "Company": ["AAPL", "AMZN", "NFLX"],
    "Standard Deviation": [
        aapl_return.std(),
        amzn_return.std(),
        nflx_return.std()
    ]
})

volatility
```

1. Which stock has the highest volatility?

NFLX has the highest volatility with a standard deviation of 0.018589.

3. What does high volatility indicate about a stock?

High volatility indicates greater price fluctuations and higher risk.

```python
closing_prices = pd.DataFrame({
    "AAPL": aapl["Close"]["AAPL"],
    "AMZN": amzn["Close"]["AMZN"],
    "NFLX": nflx["Close"]["NFLX"]
})

closing_prices.head()
```

```python
correlation_matrix = closing_prices.corr(method="pearson") #pearson corr bata hai agr ek ky price up ho rahy to kiya dusry ky bi ushi direction mein up ho rahy

correlation_matrix
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True, #Heatmap ke boxes ke andar correlation ki numerical values show karta hai.
    fmt=".2f" #value ko 2 decimal place mein convert karta hai
)

plt.title("Correlation Matrix of AAPL, AMZN and NFLX")
plt.show()
```

1. Which companies have the strongest relationship?

AMZN and NFLX have the strongest positive correlation (0.90).

2. Which companies have the weakest relationship?

AAPL and AMZN have the weakest relationship, with a correlation value of 0.65.

3. What does a high correlation value indicate?

A high positive correlation indicates that two stocks tend to move in the same direction. When one stock's price increases, the other tends to increase as well, and when one decreases, the other tends to decrease.
